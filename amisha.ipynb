----------------------------------------DATA PREPROCESSING--------------------------------



```python
import pandas as pd

df = pd.read_csv("/content/drive/MyDrive/data.csv", header=1)

print(df)

df.to_csv("/content/drive/MyDrive/data2.csv")
```

```python
df.describe()
```

```python
df = df.fillna(df.mean(numeric_only=True))
print(df.to_string())
```

```python
df = df.rename(columns={'GENDER ':"GENDER","SOFTWARE ENGINEER": "SWE", 'COMPUTER NETWORK ':'CN', 'MACHINE LEARNING ':"ML", "DATA WAREHOUSE ":"DW", "GRADE ":"GRADE", "RESULT ":"RESULT" })
print(list(df.columns))
df['GENDER'] = df['GENDER'].replace(["MALE", "male", "m", "Male", "MALE ", "Male ", "MaLe", "Mens"], "M")
df['GENDER'] = df['GENDER'].replace(["FEMALE", "female", "f", "Female", "FEMALE ", "Female "], "F")

import numpy as np
df.loc[df["GENDER"].isna(), "GENDER"] = np.random.choice(["M", "F"])
print(df.to_string())
```

```python

df['AGE'] = df['AGE'].replace(["Nineteen"], 19)
df["AGE"] = df["AGE"].fillna(0)
df['AGE'] = df["AGE"].astype(int)

age_sum = df["AGE"].sum()
age_count = df["AGE"].notna().sum()

age_mean = age_sum/age_count
age_std = df["AGE"].std()

age_lb = age_mean-age_std
age_ub = age_mean+age_std

print(age_lb, age_ub)
df = df[(df['AGE'] >= age_lb) & (df['AGE'] <= age_ub)].copy()
print(df.to_string())
```

```python
df["SWE"] = df["SWE"].astype(int)
df["TOC"] = df["TOC"].astype(int)
df["CN"] = df["CN"].astype(int)
df["ML"] = df["ML"].astype(int)
df["DW"] = df["DW"].astype(int)
print(df.to_string())
```

```python
df = df.drop(columns="GRADE")
```

```python

df["GRADE"] = df[["SWE","CN", "ML", "DW","TOC"]].mean(axis=1)
def grade(x):
  if x >= 90:
    return "O"
  if x >= 80:
    return "A"
  if x >= 70:
    return "B"
  if x >= 60:
    return "C"
  if x >= 50:
    return "D"
  if x >= 40:
    return "E"
  else:
    return "F"

df["GRADE"] = df["GRADE"].apply(grade)
print(df.to_string())
```

```python
df = df.drop(columns="RESULT")
```

```python

def result(x):
  if x == "F":
    return "FAIL"
  else:
    return "PASS"
df["RESULT"] = df["GRADE"].apply(result)
print(df.to_string())
```

----------------------------------------------------------------KNN-----------------------------------------------------------------


```python
df['SWE'] = (df['SWE'] - min(df['SWE']))/(max(df['SWE'])-min(df['SWE']))
df['TOC'] = (df['TOC'] - min(df['TOC']))/(max(df['TOC'])-min(df['TOC']))
df['CN'] = (df['CN'] - min(df['CN']))/(max(df['CN'])-min(df['CN']))
df['ML'] = (df['ML'] - min(df['ML']))/(max(df['ML'])-min(df['ML']))
df['DW'] = (df['DW'] - min(df['DW']))/(max(df['DW'])-min(df['DW']))

print(df.to_string())
```

```python
import math

train = df[:60:]
test = df[60::]

y = test["GRADE"]

predict = []

for i in range(25):
  neighbors = []
  for j in range(60):

    diff_swe = test.iloc[i]['SWE'] - train.iloc[j]['SWE']
    diff_toc = test.iloc[i]['TOC'] - train.iloc[j]['TOC']
    diff_cn = test.iloc[i]['CN'] - train.iloc[j]['CN']
    diff_ml = test.iloc[i]['ML'] - train.iloc[j]['ML']
    diff_dw = test.iloc[i]['DW'] - train.iloc[j]['DW']

    final = math.sqrt(diff_swe**2 + diff_toc**2 + diff_cn**2 + diff_ml**2 + diff_dw**2)
    neighbors.append((final, train.iloc[j]["GRADE"]))

    if len(neighbors) > 8:
      neighbors.remove(max(neighbors))

  grade = [x[1] for x in neighbors]
  prediction = max(set(grade), key=grade.count)
  predict.append(prediction)

hit = 0
miss = 0
for i in range(25):
  if predict[i] == y.iloc[i]:
    hit += 1
  else:
    miss += 1

print(hit/(hit+miss)*100)

for i in neighbors:
  print(i)

inp = []
for i in range(5):
  x = int(input("Enter Marks: "))
  inp.append(x)
```
